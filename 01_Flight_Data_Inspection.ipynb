```python
high_risk_propagation = df_clean[
    df_clean["ValidAircraftConnection"] &
    df_clean["PreviousDelayBucket"].isin([
        "31-60 min",
        "61-120 min",
        "120+ min"
    ]) &
    df_clean["TurnaroundBucket"].isin([
        "0-30 min",
        "31-60 min"
    ]) &
    df_clean["DepDel15"].notna()
]
```

```python
airline_propagation = (
    high_risk_propagation
    .groupby("Reporting_Airline")["DepDel15"]
    .agg(
        flights="count",
        delayed_departures="sum",
        departure_delay_rate="mean"
    )
    .reset_index()
)

airline_propagation.sort_values(
    "departure_delay_rate",
    ascending=False
)
```

```python
airline_baseline = (
    df_clean[
        df_clean["DepDel15"].notna()
    ]
    .groupby("Reporting_Airline")["DepDel15"]
    .agg(
        total_flights="count",
        delayed_departures="sum",
        baseline_delay_rate="mean"
    )
    .reset_index()
)
```

```python
airline_comparison = airline_propagation.merge(
    airline_baseline,
    on="Reporting_Airline"
)
```

```python
airline_comparison["PropagationLift"] = (
    airline_comparison["departure_delay_rate"]
    - airline_comparison["baseline_delay_rate"]
)
```

```python
airline_comparison.sort_values(
    "PropagationLift",
    ascending=False
)
```
